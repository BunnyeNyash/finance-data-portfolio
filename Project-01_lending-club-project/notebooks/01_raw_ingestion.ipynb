## What I Observed


> Total columns / rows in the sample:
- 150,000 rows × 151 columns



> Columns that are mostly empty (candidates to drop):
- Out of the `151` columns that are in the dataset, `73` of them have values missing. Out of these 73 columns, there are `38` columns that are more than `90%` empty.
- Upon investigation, I found out that most of these empty columns belonged to these `three` groups:
   1. **Second applicant information:** sec_app_*, revol_bal_joint, dti_joint, etc.
   2. **Hardship information:** hardship_type, hardship_reason, hardship_status, etc.
   3. **Debt settlement information:** settlement_status, settlement_amount, etc.
  
- The verdict was that the Second applicant information columns were mostly empty because very few loans in the dataset have a second applicant, the Hardship information ones because only a small number of borrowers needed or joined a hardship program, and the Debt settlement information ones because only a small number of loans went through debt settlement.
- I decided to drop these 38 columns in Notebook 2, not because the data is bad or incorrect, but simply because they are not useful for our analysis. They contain information for only a very small number of loans, and keeping them would add unnecessary columns without giving us much useful information.



> How `loan_status` breaks down, and how I'll collapse it into Good/Bad in notebook 2:
- The loan_status column tells us what happened to each loan, and here is the breakdown:
    1. **Fully Paid**: `105,691 loans` (70.46%) → the loan was successfully paid back.
    2. **Charged Off**: `26,330 loans` (17.55%) → the loan was not fully repaid.
    3. **Current**: `16,982 loans` (11.32%) → the loan is still ongoing.
    4. **Late (31–120 days**): `611 loans` (0.41%) → the borrower is seriously late with payments.
    5. **In Grace Period**: `271 loans` (0.18%) → the loan is temporarily within a payment grace period.
    6. **Late (16–30 days)**: `114 loans` (0.08%) → the borrower is late with payments.
    7. **Default**: `1 loan` (0.00%) → the loan has officially defaulted.

- To make the analysis easier, I'll group the loans into three simple categories:
    1. **Good:**
        - Fully Paid

    2. **Bad:** (all loans that failed to be repaid on schedule)
       - Charged Off
       - Default
       - Late (31–120 days)
       - Late (16–30 days)

    3. **Still Active**
       - Current
       - In Grace Period

- I will be keeping **Current** and **In Grace Period** separate because these loans are not finished yet, and we don't know whether they will eventually be paid successfully or become bad loans. Putting them into either **Good** or **Bad** would mean that we are already guessing the final outcome.
- For the main analysis in Notebook 3, I'll therefore focus on loans with a clear outcome: **Good vs. Bad**, while leaving the **Still Active** loans out of that comparison.



> Data leakage risk:
- Some columns tell us what happened after the borrower received the loan.
   1. **total_pymnt** → how much the borrower eventually paid
   2. **total_rec_prncp** → how much of the original loan was repaid
   3. **total_rec_int** → how much interest was paid
   4. **recoveries** → money recovered after a loan went bad
   5. **last_pymnt_d** → date of the last payment
   6. **last_pymnt_amnt** → amount of the last payment
   7. **out_prncp** → amount of the loan that was still outstanding

- I am planning on not using these columns in the main analysis because they are the results of the loan, and not information we had when the loan was first given.
- **For example**, if we are trying to understand why a loan became `Bad`, using `total_pymnt` would not be fair because we already know how much the borrower paid. That information comes after the loan's performance has happened, and using it is like trying to predict whether a student will pass an exam by using their final exam score, but the answer is already revealed.
- These columns will be removed from the performance analysis, to make sure our analysis focuses on the original characteristics of the borrower and loan, rather than information that became available afterward. I'll only use them when needed to understand or confirm the final loan outcome.
